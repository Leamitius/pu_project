```python
import pandas as pd
import json

# Eingabe-CSV
INPUT_FILE = "Country_codes.csv"

# Ausgabe-Dateien
OUTPUT_JSON = "countries_by_iso2.json"
OUTPUT_CSV = "countries_by_iso2.csv"

# CSV laden
df = pd.read_csv(INPUT_FILE)

# Nur gewünschte Sprachen behalten
languages = ["de", "en", "es", "fr"]
df = df[df["Language"].isin(languages)]

# Datenstruktur aufbauen
result = {}

for _, row in df.iterrows():
    iso2 = row["Alpha2Code"]
    iso3 = row["Alpha3Code"]
    lang = row["Language"]
    name = row["Name"]

    if iso2 not in result:
        result[iso2] = {
            "alpha3": iso3,
            "names": {}
        }

    result[iso2]["names"][lang] = name

# ---------- JSON speichern (empfohlen für JS) ----------
with open(OUTPUT_JSON, "w", encoding="utf-8") as f:
    json.dump(result, f, ensure_ascii=False, indent=2)

# ---------- Alternative: kompakte CSV ----------
csv_rows = []

for iso2, data in result.items():
    row = {
        "Alpha2Code": iso2,
        "Alpha3Code": data["alpha3"],
        "de": data["names"].get("de", ""),
        "en": data["names"].get("en", ""),
        "es": data["names"].get("es", ""),
        "fr": data["names"].get("fr", "")
    }
    csv_rows.append(row)

pd.DataFrame(csv_rows).to_csv(OUTPUT_CSV, index=False, encoding="utf-8")

print("Fertig!")
print(f"- JSON: {OUTPUT_JSON}")
print(f"- CSV:  {OUTPUT_CSV}")
```
